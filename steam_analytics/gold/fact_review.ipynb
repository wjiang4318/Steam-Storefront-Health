```python
%sql
USE CATALOG workspace;
USE SCHEMA default;
```

```python
%sql
CREATE OR REPLACE TABLE fact_reviews AS
SELECT
  g.game_id,
  CASE WHEN r.recommendation = 'Recommended' THEN 1 ELSE 0 END AS recommended,
  r.hours_played,
  r.helpful,
  r.funny,
  r.review_text,
  r.review_date,
  r.review_year,
  r.has_date,
  r.username,
  r.reviewer_products
FROM silver_reviews r
INNER JOIN dim_games g
  ON r.game_key = g.game_key

-- INNER JOIN reviews to dim_games on game_key to look up each game's surrogate game_id.
-- fact carries game_id (FK -> dim_games) + measures + degenerate attributes.
-- INNER = reviews for games not in dim_games are dropped (coverage scoping).y
```

```python
%sql
SELECT
  (SELECT COUNT(*) FROM silver_reviews) AS before,
  (SELECT COUNT(*) FROM fact_reviews)   AS after,
  (SELECT COUNT(*) FROM silver_reviews) - (SELECT COUNT(*) FROM fact_reviews) AS dropped
```
