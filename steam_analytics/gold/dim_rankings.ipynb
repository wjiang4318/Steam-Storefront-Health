```python
%sql
USE CATALOG workspace;
USE SCHEMA default;
```

```python
%sql
CREATE OR REPLACE TABLE dim_rankings AS
SELECT
  g.game_id,
  r.genre,
  r.rank_type,
  r.rank
FROM silver_rankings r
INNER JOIN dim_games g
  ON r.game_key = g.game_key
```

```python
%sql
SELECT
  (SELECT COUNT(*) FROM silver_rankings) AS before,
  (SELECT COUNT(*) FROM dim_rankings)    AS after,
  (SELECT COUNT(*) FROM silver_rankings) - (SELECT COUNT(*) FROM dim_rankings) AS dropped
FROM (SELECT 1)
```
