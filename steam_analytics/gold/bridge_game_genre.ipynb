```python
%sql
CREATE OR REPLACE TABLE bridge_game_genre AS
SELECT
  game_id,
  EXPLODE(genres) AS genre
FROM dim_games
```

```python
%sql
SELECT
  (SELECT COUNT(*) FROM dim_games) AS games,
  (SELECT COUNT(*) FROM bridge_game_genre) AS bridge_rows,
  (SELECT COUNT(DISTINCT game_id) FROM bridge_game_genre) AS games_in_bridge,
  (SELECT COUNT(DISTINCT genre) FROM bridge_game_genre) AS distinct_genres
FROM (SELECT 1)
```
