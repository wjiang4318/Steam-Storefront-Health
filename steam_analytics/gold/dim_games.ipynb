```python
%sql
USE CATALOG workspace;
USE SCHEMA default;
```

```python
%sql
CREATE OR REPLACE TABLE dim_games AS
SELECT
  ROW_NUMBER() OVER (ORDER BY game_key) AS game_id,
  game_key,
  game_name,
  genres,
  release_date,
  developer,
  publisher,
  rating_score,
  rating_label,
  reviews_purchased,
  reviews_english,
  review_count_inconsistent,
  link
FROM silver_games
```

```python
%sql
SELECT COUNT(*) AS rows, COUNT(DISTINCT game_id) AS ids, COUNT(DISTINCT game_key) AS keys
FROM dim_games
```
